```python
import pickle
import os
import ipywidgets as widgets
import matplotlib.pyplot as plt
from IPython.display import display, clear_output
import numpy as np

# 让 Matplotlib 在 Jupyter Notebook 内嵌显示
%matplotlib inline  

# 你的pkl文件路径（假设它们存放在一个文件夹中）
pkl_folder = "test_data"

# 读取所有pkl文件
pkl_files = [os.path.join(pkl_folder, f) for f in os.listdir(pkl_folder) if f.endswith('.pkl')]

# 加载所有数据
all_data = []
for pkl_file in pkl_files:
    with open(pkl_file, "rb") as f:
        transitions = pickle.load(f)  # 假设存储的是一个列表
        all_data.extend(transitions)  # 合并所有 transition 记录

# 获取数据总数
num_frames = len(all_data)

# 创建滑动条
slider = widgets.IntSlider(min=0, max=num_frames-1, step=1, value=0, description="Frame")

# 创建图像显示区域
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

def update_plot(frame_idx):
    """更新图像"""
    clear_output(wait=True)  # 清除上次的输出，避免重复显示
    display(slider)  # 重新显示滑动条

    transition = all_data[frame_idx]  # 取出当前帧的数据
    obs = transition["observations"]

    # 解析前置和侧面相机的RGB图像
    front_image = obs["front_camera"]
    side_image = obs["side_camera"]

    # 如果图像是灰度或者通道不匹配，进行调整
    if len(front_image.shape) == 2:
        front_image = np.stack([front_image] * 3, axis=-1)  # 转换成3通道
    if len(side_image.shape) == 2:
        side_image = np.stack([side_image] * 3, axis=-1)

    # 更新前置相机图像
    fig, axes = plt.subplots(1, 2, figsize=(10, 5))  # 重新创建画布
    axes[0].imshow(front_image)
    axes[0].set_title("Front Camera")
    axes[0].axis("off")

    # 更新侧面相机图像
    axes[1].imshow(side_image)
    axes[1].set_title("Side Camera")
    axes[1].axis("off")

    plt.show()  # 显示图像

# 绑定滑动条事件
widgets.interactive(update_plot, frame_idx=slider)

# 显示控件
display(slider)
update_plot(0)  # 显示第一帧
```
